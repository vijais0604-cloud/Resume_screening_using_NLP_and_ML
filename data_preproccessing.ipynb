## Cleaning the columns

```python
dataset.columns = dataset.columns.str.strip()

dataset.columns = dataset.columns.str.replace(r"[﻿1./]", "", regex=True)

dataset.columns = dataset.columns.str.replace(" ","", regex=True)

dataset.columns = dataset.columns.str.lower()
```

```python
drop_cols = [
    "online_links",
    "extra_curricular_organization_names",
    "extra_curricular_organization_links",
    "address",
    "company_urls",
    "issue_dates",
    "expiry_dates",
    "locations",
]

dataset = dataset.drop(columns=drop_cols)    
```

```python
dataset.columns.tolist()
```

```python
dataset = dataset.fillna("")
```

```python
import re

def clean_text(text):

    text = str(text)

    text = re.sub(r"\[|\]|'", " ", text)

    text = re.sub(r"\bnone\b", " ", text)

    text = re.sub(r"[^a-zA-Z ]", " ", text)

    text = re.sub(r"\s+", " ", text)

    return text.strip().lower()
```

```python
dataset = dataset.apply(
    lambda col: col.map(clean_text)
)
```

```python
dataset = dataset.replace(r"\bnone\b", "", regex=True)
```

```python
resume = dataset[['career_objective',
 'skills',
 'educational_institution_name',
 'degree_names',
 'passing_years',
 'educational_results',
 'result_types',
 'major_field_of_studies',
 'professional_company_names',
 'start_dates',
 'end_dates',
 'related_skils_in_job',
 'positions',
 'responsibilities',
 'extra_curricular_activity_types',
 'role_positions',
 'languages',
 'proficiency_levels',
 'certification_providers',
 'certification_skills',
]]
```

## Text Preprocessing

```python
import nltk
nltk.download("punkt")
nltk.download("stopwords")
nltk.download("wordnet")
nltk.download("punkt_tab")
```

```python
resume["resume_text"] = resume.astype(str).agg(" ".join, axis=1)
```

```python
job["job_text"] = job.astype(str).agg(" ".join, axis=1)
```

```python
from nltk.tokenize import word_tokenize,sent_tokenize
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
```

```python
def preprocess_text(text):
    sentences = sent_tokenize(text)
    for sentence in sentences:
        words = word_tokenize(sentence)
        # Further preprocessing steps can be added here
        words = [word for word in words if word not in stopwords.words("english")]
        words = [WordNetLemmatizer().lemmatize(word) for word in words]
        sentences = " ".join(words)
    return sentences    
```

```python
resume["resume_text_processed"]= resume["resume_text"].apply(preprocess_text)
job["job_text_processed"]= job["job_text"].apply(preprocess_text)
```

```python
resume_data=resume[["resume_text_processed"]]
job_data=job[["job_text_processed"]]
```

```python
resume_data.to_csv("data/resume_data_processed.csv", index=False)
job_data.to_csv("data/job_data_processed.csv", index=False)
```
